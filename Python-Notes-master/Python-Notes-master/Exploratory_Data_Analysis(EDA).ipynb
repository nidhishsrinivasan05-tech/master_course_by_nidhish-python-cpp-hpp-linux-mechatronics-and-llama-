```python
# Install necessary libraries
!pip install cufflinks plotly
# Import necessary libraries
import pandas as pd
import seaborn as sns
import cufflinks as cf
import plotly.offline as pyo
from plotly.io import to_html
from IPython.display import display, HTML
# Set up Plotly offline mode
pyo.init_notebook_mode(connected=True)
# Enable Cufflinks
cf.go_offline()
# Load a built-in dataset
df = sns.load_dataset('tips')
# Create interactive plots with Cufflinks
# Scatter Plot
scatter_fig = df[['total_bill', 'tip']].iplot(kind='scatter', x='total_bill', y='tip', mode='markers', title='Total Bill vs Tip', asFigure=True)
display(HTML(to_html(scatter_fig)))
# Bar Chart
bar_fig = df.groupby('day').mean()['total_bill'].iplot(kind='bar', title='Average Total Bill by Day', asFigure=True)
display(HTML(to_html(bar_fig)))
# Box Plot
box_fig = df.iplot(kind='box', y='total_bill', title='Box Plot of Total Bill', asFigure=True)
display(HTML(to_html(box_fig)))
# Line Chart
line_fig = df.groupby('day').mean()['total_bill'].iplot(kind='line', title='Average Total Bill by Day', asFigure=True)
display(HTML(to_html(line_fig)))
```

```python
# Import the required libraries
import vizard
import pandas as pd

# Sample DataFrame for a regression problem
data = {
    'age': [25, 35, 45, 55, 65],
    'salary': [50000, 60000, 70000, 80000, 90000],
    'experience': [1, 5, 10, 15, 20],
    'education_level': ['Bachelors', 'Masters', 'PhD', 'Bachelors', 'Masters'],
    'salary_next_year': [52000, 64000, 71000, 82000, 95000]  # Target variable
}
df = pd.DataFrame(data)

# Config class setup for a regression problem
class config:
    PROBLEM_TYPE = 'regression'
    DEPENDENT_VARIABLE = 'salary_next_year'
    CATEGORICAL_INDEPENDENT_VARIABLES = ['education_level']
    CONTINUOUS_INDEPENDENT_VARIABLES = ['age', 'salary', 'experience']
    TEXT_VARIABLES = []

# Create the Vizard object for normal plots
viz = vizard.Vizard(df, config)

# For interactive plots
viz_in = vizard.VizardIn(df, config)

# Now you can generate plots, e.g., for regression analysis
# In the interactive setting (viz_in), you'll get interactivity.
```

```python
viz.check_missing()
```

```python
viz.count_missing()
```

```python
viz.count_unique()
```

```python
viz.dependent_variable()
```

```python
viz = vizard.Vizard(df, config)
viz2pdf = vizard.Viz2PDF('viz_report.pdf')

plots = [
    viz.check_missing(),
    viz.count_missing(),
    viz.count_unique(),
    viz.dependent_variable(),
    viz.categorical_variables(),
    viz.continuous_variables(),
    viz.pairwise_scatter(),
    viz.pairwise_violin(),
    viz.pairwise_crosstabs(),
]
viz2pdf(plots)
```

```python
import altair as alt
from vega_datasets import data
# Load the built-in 'cars' dataset
cars = data.cars()
# Basic scatter plot of Horsepower vs. Miles per Gallon
scatter = alt.Chart(cars).mark_point().encode(
 x='Horsepower',
 y='Miles_per_Gallon',
 color='Origin'
).properties(
 title='Horsepower vs. MPG by Origin'
)
scatter
```

**1. Line Chart**

```python
# Average horsepower over time
line_chart = alt.Chart(cars).mark_line().encode(
 x='Year:T',
 y='average(Horsepower)',
 color='Origin'
).properties(
 title='Average Horsepower Over Time by Origin'
)
line_chart
```

**2. Bar Chart**

```python
# Count of cars by origin
bar_chart = alt.Chart(cars).mark_bar().encode(
 x='count()',
 y='Origin',
 color='Origin'
).properties(
 title='Number of Cars by Origin'
)
```

```python
bar_chart
```

**3. Histogram**

```python
# Histogram of MPG
histogram = alt.Chart(cars).mark_bar().encode(
 alt.X("Miles_per_Gallon:Q", bin=True),
 y='count()'
).properties(
 title='Distribution of Miles per Gallon (MPG)'
)
histogram
```

**4. Heatmap**

```python
# Heatmap of Horsepower vs. Weight
heatmap = alt.Chart(cars).mark_rect().encode(
 x='Horsepower:Q',
 y='Weight_in_lbs:Q',
 color='count()'
).properties(
 title='Heatmap of Horsepower vs. Weight'
)
heatmap
```

**Adding Tooltips**

```python
scatter_tooltip = alt.Chart(cars).mark_point().encode(
 x='Horsepower',
 y='Miles_per_Gallon',
 color='Origin',
 tooltip=['Name', 'Horsepower', 'Miles_per_Gallon']
).properties(
 title='Horsepower vs. MPG with Tooltips'
)
scatter_tooltip
```

**Creating Linked Plots**


```python
# Define selection
brush = alt.selection_interval()
# Scatter plot with selection
scatter_select = alt.Chart(cars).mark_point().encode(
 x='Horsepower',
 y='Miles_per_Gallon',
 color=alt.condition(brush, 'Origin', alt.value('lightgray'))
).add_selection(
 brush
)
# Histogram that reflects the selection
hist_select = alt.Chart(cars).mark_bar().encode(
 x='Origin',
 y='count()',
 color='Origin'
).transform_filter(
 brush
)
scatter_select | hist_select
```

Customizing Altair Charts

```python
# Customizing a scatter plot with size, theme, and labels
alt.themes.enable('dark')
scatter_custom = alt.Chart(cars).mark_point(size=60).encode(
 x='Horsepower',
 y='Miles_per_Gallon',
 color='Origin',
 text=alt.Text('Name')
).properties(
 title='Customized Scatter Plot',
 width=500,
 height=400
)
scatter_custom
```

```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt

# Sample text data
text = """Python is a high-level, general-purpose programming language. Its design philosophy
emphasizes code readability with the use of significant indentation. Python is dynamically-typed
and garbage-collected. It supports multiple programming paradigms, including structured, object-oriented,
and functional programming."""

# Define stopwords
stopwords = set(STOPWORDS)

# Generate the word cloud
wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white',
    stopwords=stopwords,
    min_font_size=10
).generate(text)

# Display the word cloud
plt.figure(figsize=(10, 5), facecolor=None)
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis("off")  # No axes for a clean look
plt.tight_layout(pad=0)
plt.show()
```
